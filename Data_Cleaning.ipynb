```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("ecommerce_customer_churn_dataset.csv", low_memory=False)
df.head()
```

Identify the number of missing values in each column before cleaning.

```python
df.isnull().sum()
```

Since all missing values are present in numerical columns, replace them with the median value of each column.
The median is preferred because it is less affected by extreme values (outliers) than the mean.

```python
numeric_columns = df.select_dtypes(include=np.number).columns
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isna().sum()
```
